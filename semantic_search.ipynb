```python
import json
import numpy as np
import pandas as pd
from sentence_transformers import SentenceTransformer, util

with open('code_corpus.json', 'r', encoding='utf-8') as f:
    corpus = json.load(f)

with open('eval_questions.json', 'r', encoding='utf-8') as f:
    questions = json.load(f)

with open('categories.json', 'r', encoding='utf-8') as f:
    categories = json.load(f)

print(f"Фрагментов кода: {len(corpus)}")
print(f"Вопросов: {len(questions)}")
print("\nПример фрагмента кода:")
print(json.dumps(corpus[0], ensure_ascii=False, indent=2)[:400])
```

```python
import time

texts = [f"{item['description']}\n{item['code']}" for item in corpus]

model_names = [
    "paraphrase-multilingual-MiniLM-L12-v2",
    "paraphrase-multilingual-mpnet-base-v2"
]

embeddings = {}

for name in model_names:
    print(f"\n=== Загрузка модели: {name} ===")
    start = time.time()
    
    model = SentenceTransformer(name, device="cpu")
    emb = model.encode(texts, show_progress_bar=True, convert_to_numpy=True)
    
    embeddings[name] = emb
    elapsed = time.time() - start
    print(f"Готово за {elapsed:.1f} сек. Размерность: {emb.shape}")
```
